```python
import numpy as np 
```

# Linear Regression Hypothesis
- y^= b0 + b1 * x
- y^ (predicted value)
- b0 (bias / y-intercept)
- b1 (weight / slope)
- x (independent variable / feature)
- error e = (y - y^)


```python
x=np.array([1,2,3,4,5])
y=np.array([5,7,9,11,13])

```

# simple error 

```python
# Simple Error (Sum of Errors)

# J(b0​,b1​)=∑​(yi​−y^​i​)

b0,b1=0 ,0
alpha=0.01
terms = 1000

for i in range(terms):
    y_pred = b0 + b1 * x
    e = y - y_pred
    b0+= alpha*np.sum(e)
    b1+= alpha*np.sum(e*x)
    
print(b0,b1)
    
```

# Absolute Error

```python
# Absolute Error (Sum of Absolute Errors)

# J(b0​,b1​)=∑​|yi​−y^​i​|

b0,b1=0 ,0
alpha=0.01
terms = 1000

for i in range(terms):
    y_pred = b0 + b1 * x
    e = y - y_pred
    b0+= alpha*np.sum(np.sign(e))
    b1+= alpha*np.sum(np.sign(e)*x) 
    
print(b0,b1)
```

# square Error


```python
# square Error (Sum of Squared Errors)  
# J(b0​,b1​)=∑​(yi​−y^​i​)2

b0,b1=0 ,0
alpha=0.01
terms = 1000

for i in range(terms):
    y_pred = b0 + b1 * x
    e = y - y_pred
    b0 += alpha * (2 * e).sum() / len(y)
    b1 += alpha * (2 * (e * x)).sum() / len(y)
    
print(b0,b1)
```
