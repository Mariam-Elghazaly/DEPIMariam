```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

# Basic graphs 

```python
plt.plot([1,2,3],[4,5,6])
```

```python
# we recommend : 
x=[0,1,2,3,4]
y=[0,2,4,6,8]
plt.plot(x,y,label='2x', color='red',linewidth=2, marker='x',markersize=10,markeredgecolor='blue',linestyle='--')
plt.title('garph',fontdict={'fontname':"Comic Sans MS", "fontsize":20})
plt.xlabel("state",fontdict={'fontname':"Comic Sans MS", "fontsize":20})
plt.ylabel('temp',fontdict={'fontname':"Comic Sans MS", "fontsize":20})
plt.xticks([0,1,2,3])# we automatic label the graph
plt.yticks([0,2,4,6,8,10])
plt.legend()
plt.show()

```

```python
# we can have a short hand 


x=[0,1,2,3,4]
y=[0,2,4,6,8]
#plt.plot(x,y,label='2x', color='red',linewidth=2, marker='x',markersize=10,markeredgecolor='blue',linestyle='--')
# color marker line
plt.plot(x,y,'b^--',label='2x')


plt.title('garph',fontdict={'fontname':"Comic Sans MS", "fontsize":20})
plt.xlabel("state",fontdict={'fontname':"Comic Sans MS", "fontsize":20})
plt.ylabel('temp',fontdict={'fontname':"Comic Sans MS", "fontsize":20})
plt.xticks([0,1,2,3])# we automatic label the graph
plt.yticks([0,2,4,6,8,10])
plt.legend()
plt.show()
```
